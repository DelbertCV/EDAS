### Missing values & null values

```python
# Missing values function: it receives a DataFrame and outputs another DataFrame

def missing_values_report(df: pd.DataFrame) -> pd.DataFrame:

    return (
        pd.DataFrame({
            "Count of missing values": df.isna().sum(),
            "Percent of missing values": (df.isna().sum() / len(df) * 100).round(1),
        })
    )
```

```python
missing_values_report(train_df)
```
